```python
# ==========================
# Import Libraries
# ==========================
import pandas as pd
import numpy as np
import pickle

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

# ==========================
# Load Dataset
# ==========================
df = pd.read_csv("flood.csv")

print("Shape:", df.shape)

# ==========================
# Input / Output
# ==========================
X = df.drop('FloodProbability', axis=1)
y = df['FloodProbability']

feature_names = X.columns.tolist()

# ==========================
# Train Test Split
# ==========================
Xtrain, Xtest, ytrain, ytest = train_test_split(
    X, y,
    test_size=0.2,
    random_state=21
)

# ==========================
# Model
# ==========================
model = RandomForestRegressor(
    n_estimators=300,
    max_depth=15,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

# ==========================
# Train
# ==========================
model.fit(Xtrain, ytrain)

# ==========================
# Predictions
# ==========================
train_pred = model.predict(Xtrain)
test_pred = model.predict(Xtest)

# ==========================
# Evaluation
# ==========================
print("Train R2:", r2_score(ytrain, train_pred))
print("Test R2 :", r2_score(ytest, test_pred))
print("MAE:", mean_absolute_error(ytest, test_pred))
print("MSE:", mean_squared_error(ytest, test_pred))
print("RMSE:", np.sqrt(mean_squared_error(ytest, test_pred)))

# ==========================
# Feature Importance
# ==========================
importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": model.feature_importances_
}).sort_values(by="Importance", ascending=False)

print("\nTop Features:\n", importance)

# ==========================
# SAVE MODEL (IMPORTANT FIX)
# ==========================
model_package = {
    "model": model,
    "features": feature_names
}

with open("model.pkl", "wb") as file:
    pickle.dump(model_package, file)

print("\n✅ Model saved successfully as model.pkl")
```
